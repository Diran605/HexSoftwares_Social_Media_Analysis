```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df= pd.read_csv('social_media_dataset.csv')
```

```python
df
```

```python
#platform wise analysis
#country-wise analysis
#engagement analysis(views, likes shares comments)
#platform-wise engagement
#Region-wise Popular Content-type
```

```python
df['platform'].value_counts().plot(kind="pie", autopct="%1.1f%%")
plt.title("Most Popular Platform")
```

```python
ax= df.groupby("region")["platform"].value_counts().unstack().plot(kind="bar", stacked=True, figsize=(10,8))
plt.title("Platform used in all country")
plt.xlabel('Country')
[ax.bar_label(container, fmt="{:,.0f}") for container in ax.containers]
plt.ylabel('Sum of Used')
plt.show()
```

```python
ax= df.groupby('region')[['views', 'likes', 'shares', 'comments']].sum().plot(kind="bar", figsize=(15,6))
plt.xlabel('country')
[ax.bar_label(container, fmt="{:,.0f}") for container in ax.containers]
plt.ylabel('count')
ax.set_yscale("log")
plt.title('Counry-wise Engagement')
plt.show()
```

```python
ax= df.groupby('platform')[['views', 'likes', 'shares', 'comments']].sum().plot(kind="bar", figsize=(15,6))
plt.xlabel('platform')
[ax.bar_label(container, fmt="{:,.0f}") for container in ax.containers]
plt.ylabel('count')
ax.set_yscale("log")
plt.title('platform-wise Engagement')
plt.show()
```

```python
ax =df.groupby("region")['content_type'].value_counts().unstack().plot(kind="bar", stacked=True, figsize=(15,6))
plt.xlabel('country')
[ax.bar_label(container, fmt="{:,.0f}") for container in ax.containers]
plt.title("Most Popular Content Type in All country")
plt.show()
```

```python
region_platform_counts=df.groupby('region')['platform'].value_counts().unstack()

for region in region_platform_counts.index:
    plt.figure(figsize=(8,8))
    plt.pie(region_platform_counts.loc[region], labels= region_platform_counts.columns, autopct="%1.1f%%", startangle=45)
    plt.title(f"platform used in {region}")
    plt.show()
```
